Input dari Source untuk mendapatkan string atau scraping

```python
web_text = requests.get("example.com").text
# decompose (dari <script> <style> ,.. yang tidak di perlukan untuk di hilangkan ) untuk mendapatkan innertext 
soup = BeautifulSoup(web_text) 
for sp in soup(["style","scripe"]) : 
    sp.decompose()
text = " ".join(soup.stripped_strings)
print(text)
    

```

Manipulasi String , 
1. capita ke huruf kecil 
2. menghapus angka yang tidak di butuhkan 
3. merubah tanda baca 
4. hapus dari space atau karacter kosong 

```python
text = text.lower()
# penghapuskan angka  dari semua string (\d)
text = re.sub(r"\d"," ",text)
# penghapusan tanda baca
text = text.translate(str.maketrans("","",string.punctuation))
text = text.strip()
```
